```python
import duckdb
import pandas as pd  # Watch this turn from grey to active!

# 1. Start your in-memory database
conn = duckdb.connect(database=':memory:')

# 2. Run a quick system query to get database tables
# (Even though it's empty right now, it will return a structure)
df = conn.execute("SELECT * FROM duckdb_tables();").df()

# 3. This line uses the 'pd' module to inspect our data structure
print("Data Type:", type(df))
print(df)
```
